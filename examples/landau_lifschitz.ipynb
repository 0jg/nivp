# Landau-Lifschitz Equation

This notebook trains three neural networks for the Landau-Lifschitz equation and compares the result with both RK4 and the analytical solution.

```python
from dataclasses import dataclass
from pathlib import Path
import json
import os
import sys

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import torch
from mpl_toolkits.mplot3d import Axes3D
from tqdm import tqdm

_ = Axes3D
matplotlib.rcParams["font.family"] = "Times New Roman"
matplotlib.rcParams["font.size"] = 11
matplotlib.rcParams["mathtext.fontset"] = "stix"
matplotlib.rcParams["axes.linewidth"] = 0.8
matplotlib.rcParams["xtick.major.width"] = 0.8
matplotlib.rcParams["ytick.major.width"] = 0.8

cwd = Path.cwd()
EXAMPLES_DIR = cwd if cwd.name == "examples" else cwd / "examples"
PROJECT_ROOT = EXAMPLES_DIR.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from helpers import (
    DATA_TYPE,
    DEVICE,
    SimpleMLP,
    ensure_run_dirs,
    expected_path_tensor_first_order,
    first_order_dynamics,
)

print(f"Using device: {DEVICE}")
```

```python
@dataclass(frozen=True)
class LandauLifschitz:
    alpha: float = 0.1
    H0: float = 1.0
    dim: int = 3

    def time_derivative(self, magnetisation: torch.Tensor) -> torch.Tensor:
        field = torch.zeros_like(magnetisation)
        field[..., 2] = self.H0
        magnetisation_cross_field = torch.cross(magnetisation, field, dim=-1)
        damping_term = torch.cross(magnetisation, magnetisation_cross_field, dim=-1)
        return -magnetisation_cross_field - self.alpha * damping_term


def compute_analytical(time_values, alpha):
    mx = np.cosh(alpha * time_values) ** -1 * np.cos(time_values)
    my = np.cosh(alpha * time_values) ** -1 * np.sin(time_values)
    mz = np.tanh(alpha * time_values)
    return np.column_stack([mx, my, mz])


def plot_components(time_values, predicted, rk4_reference, analytical_reference):
    fig = plt.figure(figsize=(10, 5), dpi=150)
    fig.patch.set_facecolor("white")
    gs = fig.add_gridspec(3, 3, height_ratios=[0.06, 0.75, 0.25], hspace=0.15, wspace=0.25)

    ax_legend = fig.add_subplot(gs[0, :])
    ax_legend.axis("off")
    line_neural, = ax_legend.plot([], [], color="black", linewidth=2, label="Neural IVP")
    line_rk4, = ax_legend.plot([], [], color="darkgrey", linewidth=2, linestyle="--", label="RK4")
    line_analytical, = ax_legend.plot([], [], color="lightgrey", linewidth=2, label="Analytical")
    ax_legend.legend(handles=[line_neural, line_rk4, line_analytical], loc="upper center", ncol=3, frameon=False)

    component_labels = [r"$m_x(\tau)$", r"$m_y(\tau)$", r"$m_z(\tau)$"]
    for index, label in enumerate(component_labels):
        ax = fig.add_subplot(gs[1, index])
        ax.plot(time_values, analytical_reference[:, index], color="lightgrey", linewidth=2)
        ax.plot(time_values, rk4_reference[:, index], color="darkgrey", linewidth=2, linestyle="--")
        ax.plot(time_values, predicted[:, index], color="black", linewidth=2)
        ax.set_ylabel(label)
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        ax.tick_params(direction="in", which="both", left=True, bottom=True)
        ax.set_xticklabels([])

        residual_ax = fig.add_subplot(gs[2, index])
        residual_ax.plot(time_values, predicted[:, index] - analytical_reference[:, index], color="black", linewidth=2)
        residual_ax.plot(time_values, rk4_reference[:, index] - analytical_reference[:, index], color="darkgrey", linewidth=2, linestyle="--")
        residual_ax.axhline(0.0, color="black", linewidth=0.5, alpha=0.3)
        residual_ax.set_xlabel(r"$\tau$")
        residual_ax.set_ylabel("Residual")
        residual_ax.spines["top"].set_visible(False)
        residual_ax.spines["right"].set_visible(False)
        residual_ax.tick_params(direction="in", which="both", left=True, bottom=True, labelsize=9)
    return fig


def plot_phase_space(predicted, rk4_reference, analytical_reference):
    fig = plt.figure(figsize=(7, 7), dpi=150)
    fig.patch.set_facecolor("white")
    ax = fig.add_subplot(111, projection="3d")
    ax.plot(analytical_reference[:, 0], analytical_reference[:, 1], analytical_reference[:, 2], color="lightgrey", linewidth=2, label="Analytical")
    ax.plot(rk4_reference[:, 0], rk4_reference[:, 1], rk4_reference[:, 2], color="darkgrey", linewidth=2, linestyle="--", label="RK4")
    ax.plot(predicted[:, 0], predicted[:, 1], predicted[:, 2], color="black", linewidth=2, label="Neural IVP")
    ax.scatter([predicted[0, 0]], [predicted[0, 1]], [predicted[0, 2]], color="black", s=70, marker="o")
    ax.scatter([predicted[-1, 0]], [predicted[-1, 1]], [predicted[-1, 2]], color="black", s=70, marker="s")
    ax.set_xlim([-1.2, 1.2])
    ax.set_ylim([-1.2, 1.2])
    ax.set_zlim([-1.2, 1.2])
    ax.set_box_aspect([1, 1, 1])
    ax.view_init(elev=25, azim=240)
    ax.set_xlabel(r"$m_x$")
    ax.set_ylabel(r"$m_y$")
    ax.set_zlabel(r"$m_z$")
    ax.grid(False)
    ax.legend(loc="upper center", frameon=False, ncol=3, bbox_to_anchor=(0.5, 1.02))
    return fig


def plot_loss_history(loss_history):
    fig = plt.figure(figsize=(8, 5), dpi=150)
    fig.patch.set_facecolor("white")
    ax = fig.add_subplot(111)
    ax.semilogy(loss_history, linewidth=2, color="black")
    ax.set_xlabel(r"Epoch")
    ax.set_ylabel(r"Loss")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.tick_params(direction="in", which="both", left=True, bottom=True)
    ax.tick_params(labelbottom=True, labelleft=True)
    return fig
```

```python
params = {
    "activation": "sechlu",
    "hidden": 64,
    "epochs": 30_000,
    "lr": 1e-3,
    "T_MIN": 0.0,
    "T_MAX": 20.0,
    "TIMESTEP": 0.01,
    "mx0": 1.0,
    "my0": 0.0,
    "mz0": 0.0,
    "alpha": 0.25,
    "H0": 1.0,
    "ODE_REGULARISER": 1.0,
    "INITIAL_CONDITION_REGULARISER": 1.0,
    "JOB_ID": 1,
}

dt = params["TIMESTEP"]
steps = int(round((params["T_MAX"] - params["T_MIN"]) / dt)) + 1
time_grid = torch.linspace(
    params["T_MIN"],
    params["T_MAX"],
    steps,
    dtype=DATA_TYPE,
    device=DEVICE,
)
t = time_grid.unsqueeze(0).unsqueeze(2).requires_grad_(True)

init_state = torch.tensor([params["mx0"], params["my0"], params["mz0"]], dtype=DATA_TYPE, device=DEVICE)
system = LandauLifschitz(alpha=params["alpha"], H0=params["H0"])
expected = expected_path_tensor_first_order(
    system,
    init_state,
    params["T_MIN"],
    params["T_MAX"],
    dt,
)
```

```python
models = [
    SimpleMLP(1, params["hidden"], 1, params["activation"]).to(DEVICE),
    SimpleMLP(1, params["hidden"], 1, params["activation"]).to(DEVICE),
    SimpleMLP(1, params["hidden"], 1, params["activation"]).to(DEVICE),
]
optimisers = [torch.optim.Adam(model.parameters(), lr=params["lr"]) for model in models]
loss_fn = torch.nn.MSELoss()
loss_history = []


def evaluate():
    components = [model(t) for model in models]
    magnetisation = torch.cat(components, dim=2)
    _, residuals = first_order_dynamics(system, magnetisation, t)
    ic_loss = loss_fn(magnetisation[:, 0, :], init_state.unsqueeze(0))
    ode_loss = loss_fn(residuals, torch.zeros_like(residuals))
    total_loss = (
        params["ODE_REGULARISER"] * ode_loss
        + params["INITIAL_CONDITION_REGULARISER"] * ic_loss
    )
    return magnetisation, total_loss


for epoch in tqdm(range(params["epochs"])):
    magnetisation, total_loss = evaluate()
    for optimiser in optimisers:
        optimiser.zero_grad()
    total_loss.backward()
    for optimiser in optimisers:
        optimiser.step()
    loss_history.append(total_loss.item())

    if epoch % 1000 == 0 or epoch == params["epochs"] - 1:
        time_values = t[0, :, 0].detach().cpu().numpy()
        predicted_values = magnetisation[0].detach().cpu().numpy()
        rk4_values = expected[0].detach().cpu().numpy()
        analytical_values = compute_analytical(time_values, params["alpha"])
        display(plot_components(time_values, predicted_values, rk4_values, analytical_values))
        display(plot_phase_space(predicted_values, rk4_values, analytical_values))
        display(plot_loss_history(loss_history))

final_magnetisation, _ = evaluate()
```

```python
output_base = EXAMPLES_DIR / "outputs"
paths = ensure_run_dirs(str(output_base), params, system_name="landau_lifshitz")
model_paths = [
    os.path.join(paths["models"], f"{params['JOB_ID']}_mx.pth"),
    os.path.join(paths["models"], f"{params['JOB_ID']}_my.pth"),
    os.path.join(paths["models"], f"{params['JOB_ID']}_mz.pth"),
]
loss_history_path = os.path.join(paths["paths"], "loss_history.json")
components_path = os.path.join(paths["figures"], "components.png")
phase_space_path = os.path.join(paths["figures"], "phase_space_3d.png")
loss_plot_path = os.path.join(paths["figures"], "loss_history.png")

for model, path in zip(models, model_paths):
    torch.save(model, path)
with open(loss_history_path, "w") as handle:
    json.dump(loss_history, handle)

time_values = t[0, :, 0].detach().cpu().numpy()
predicted_values = final_magnetisation[0].detach().cpu().numpy()
rk4_values = expected[0].detach().cpu().numpy()
analytical_values = compute_analytical(time_values, params["alpha"])

components_figure = plot_components(time_values, predicted_values, rk4_values, analytical_values)
components_figure.savefig(components_path, dpi=150, bbox_inches="tight")
phase_figure = plot_phase_space(predicted_values, rk4_values, analytical_values)
phase_figure.savefig(phase_space_path, dpi=150, bbox_inches="tight")
loss_figure = plot_loss_history(loss_history)
loss_figure.savefig(loss_plot_path, dpi=150, bbox_inches="tight")

print("Saved:")
print(f"- Model mx: {model_paths[0]}")
print(f"- Model my: {model_paths[1]}")
print(f"- Model mz: {model_paths[2]}")
print(f"- Loss history: {loss_history_path}")
print(f"- Component figure: {components_path}")
print(f"- 3D phase space: {phase_space_path}")
print(f"- Loss plot: {loss_plot_path}")
```
